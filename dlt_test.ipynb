```python
import dlt
from dlt.sources.filesystem import filesystem, read_parquet
from dotenv import load_dotenv
import os
import polars as pl


load_dotenv()
```

```python
# define the source extraction 
filesystem_source = filesystem(
    bucket_url="s3://silver-lake/gameweeks",
    file_glob="*.parquet",
        credentials={
        "aws_access_key_id": os.getenv("MINIO_ACCESS_KEY"),
        "aws_secret_access_key": os.getenv("MINIO_SECRET_KEY"),
        "endpoint_url": f"http://{os.getenv('MINIO_ENDPOINT')}"  # just the hostname, like "minio-yokckg4o44wg40wogk0okgks.65.108.88.160.sslip.io"
    },
) | read_parquet()
```

```python
# fetches credentials from env automatically through predetermined variable names
# create a pipeline object which moves data from source to destination
pipeline = dlt.pipeline(
    pipeline_name="test_pipeline", 
    destination="duckdb",
    dataset_name="gameweeks_silver",
    )

info = pipeline.run(filesystem_source)
print(info)
```
